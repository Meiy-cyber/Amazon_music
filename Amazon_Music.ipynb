```python
from sklearn.preprocessing import StandardScaler

# 1. StandardScaler-ah initialize
scaler = StandardScaler()

# 2. Numerical columns-ah

scaled_features = scaler.fit_transform(df)


import pandas as pd
df_scaled = pd.DataFrame(scaled_features, columns=df.columns)

print("Data normalized successfully!")
```

```python
# Recommended features list-ah create panrom
selected_features = [
    'danceability', 
    'energy', 
    'loudness', 
    'speechiness', 
    'acousticness', 
    'instrumentalness', 
    'liveness', 
    'valence', 
    'tempo', 
    'duration_ms'
]


X = df[selected_features]

print("Selected features shape:", X.shape)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1. Feature Scaling (Normalize data)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 2. PCA - Reduce to 2 components for visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Original Shape:", X_scaled.shape)
print("Reduced Shape after PCA:", X_pca.shape)
```

```python
!pip install matplotlib
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# SSE (Inertia) values
sse = []
k_range = range(1, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    sse.append(kmeans.inertia_)

# Plotting the Elbow Curve
plt.figure(figsize=(8, 4))
plt.plot(k_range, sse, marker='o', linestyle='-', color='b')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('SSE (Inertia)')
plt.title('Elbow Method For Optimal k')
plt.grid(True)
plt.show()
```

```python

optimal_k = 4 

kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_scaled)

df['cluster'] = cluster_labels

print("Clustering completed! Cluster column added successfully.")
print(df['cluster'].value_counts())
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=0.5, min_samples=5)
df['dbscan_cluster'] = dbscan.fit_predict(X_scaled)

# Noise (outliers) count
print("Cluster counts (including -1 for noise):")
print(df['dbscan_cluster'].value_counts())
```

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score

# 1. Evaluation Metrics calculation
sil_score = silhouette_score(X_scaled, df['cluster'])
db_index = davies_bouldin_score(X_scaled, df['cluster'])

print(f"Silhouette Score (Higher is better): {sil_score:.4f}")
print(f"Davies-Bouldin Index (Lower means better separation): {db_index:.4f}")

# 2. Cluster Profiling (Mean values of each feature per cluster)
numeric_features = [
    'danceability', 'energy', 'loudness', 'speechiness', 
    'acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo', 'duration_ms'
]

cluster_profile = df.groupby('cluster')[numeric_features].mean()
print("\n--- Cluster Profiles (Mean Values) ---[cite: 10]")
print(cluster_profile)
```
